```r
asd_mice_0 <- asd_clean |>
  # 1. Lowercase, trim whitespace, and Title-Case all character variables to standardize categories under categorical columns
  mutate(across(
    where(is.character),
    ~ str_to_title(str_to_lower(str_trim(.x)))
  ))
```

```r
asd_mice <- asd_mice_0 %>%
  mutate(
    across(where(is.character), as.factor),
    across(
      c(a1:a9, a10_autism_spectrum_quotient),
      ~ factor(.x, levels = c(0, 1))
    )
  )

  asd_mice
```

```r
vars_to_check <- c(
  "speech_delay_language_disorder",
  "learning_disorder",
  "genetic_disorders",
  "depression",
  "global_developmental_delay_intellectual_disability",
  "anxiety_disorder",
  "sex",
  "jaundice",
  "family_mem_with_asd",
  "asd_traits"
)

lapply(
  vars_to_check,
  function(v) {
    list(
      variable = v,
      table = table(
        asd_mice$social_behavioural_issues,
        asd_mice[[v]],
        useNA = "ifany"
      )
    )
  }
)
```

Every single patient with social_behavioural_issues == Yes has depression == Yes, and every patient with social_behavioural_issues == No has depression == No. Because these two columns contain identical information across all 1,970 complete cases, mice detected linear dependency and automatically dropped social_behavioural_issues.

speech_delay_language_disorder & learning_disorder: Whenever social_behavioural_issues == Yes, 100% of patients (1,040 / 1,040) also have Yes for these conditions. There are zero No cases.

Hence, social_behavioral_issues were dropped.

Let's check whether there is any possibility of MAR in social_behavioral issues.

```r
table(
  Respondent = asd_mice$who_completed_the_test,
  Missing_Social_Issues = is.na(asd_mice$social_behavioural_issues),
  useNA = "ifany"
)  # Checking for structural missingness; If all missing values are under one category, it means that that variable can explain
   # can explain the missigness. Hence, MAR possibility.
```

All missing Social behavioral issues are explained by the respondent group - Health Care Professional. So, possibility of MAR.

```r
# 2. Age Distribution across Missing vs. Observed groups
asd_mice |>
  group_by(Missing_Social_Issues = is.na(social_behavioural_issues)) |>
  summarise(
    Count = n(),
    Min_Age = min(age_years, na.rm = TRUE),
    Median_Age = median(age_years, na.rm = TRUE),
    Max_Age = max(age_years, na.rm = TRUE),
    Mean_Age = round(mean(age_years, na.rm = TRUE), 2)
  ) # Checking for structural missingness; If all missing values are under age group, it means that that variable can explain
   # can explain the missigness. Hence, MAR possibility.
```

Age distribution is same across both groups. Could be MCAR. Overall, till now, the results point to MAR wrt respondent & social behavioral issues and asd_traits & qchat10 at least. So, there is a possibility that it is not completely MNAR. Even the collinear variable in MICE shows the possiblity of MAR. So, relying on only complete cases might not be the best even if some other variables show the possibility of MCAR. MICE implementation was justified.

However, with social_behavioral_issues being highly explained by depression and also other variables, it would be wise to remove it.

And the different autism scores are meant for different age groups.

AQ10 is meant only for 16 and older.
CARS is meant for 2 and above.
SRS is meant for 4-18 year olds.
QChat10 is meant for 18-24 month olds.
So, the scores are not standardized.

Also, they could be highly correlated to ASD traits as they are used for the screening itself. So, dropping them would be better.
