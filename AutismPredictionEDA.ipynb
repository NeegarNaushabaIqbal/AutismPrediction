The AQ-10 Autism Spectrum Quotient (AQ-10) is a quick questionnaire that primary care practitioners can use to see if a person should be referred for an autism assessment. Max score = 10. Each individual question has a score of 1.

https://embrace-autism.com/aq-10/

The Social Responsiveness Scale (SRS) is a 65-item rating scale that measures the severity of autistic symptomatology as a quantitative trait, among children clinically affected by autism spectrum conditions as well as among children in the general population. Total Score = 10.

https://link.springer.com/rwe/10.1007/978-1-4419-1698-3_296

The Q-CHAT-10, or Quantitative Checklist for Autism in Toddlers, serves as a screening tool used to identify signs of autism spectrum disorder in toddlers aged eighteen to twenty-four months.Highest score = 10.

https://www.dochub.com/fillable-form/373227-q-chat-10-quantitative-checklist-for-autism-in-toddlers


The Childhood Autism Rating Scale (CARS) is a 15-item behavioral observation tool that clinicians use to rate the presence and severity of autism spectrum disorder in children, producing a total score between 15 and 60. Each question is graded on a 1 to 4 scale.

https://neurolaunch.com/childhood-autism-rating-scale/

For all these scores, higher value means higher chances of autism.

But before that, let us estimate the possibility of Missing Completely at Random (MCAR), Missing at Random (MAR), and Missing Not at Random (MNAR) in more detail.

Missing Completely at Random (MCAR) occurs when the probability of missingness is independent of both observed and unobserved values. Think of it like equipment randomly failing — the missingness has nothing to do with the actual data values. Under MCAR, simply analyzing complete cases yields unbiased estimates, though with reduced precision due to smaller sample sizes.

Missing at Random (MAR) occurs when missingness depends only on observed data. For example, older patients may be less likely to complete lengthy questionnaires, but conditional on age, the missingness is random. MAR is more realistic than MCAR and allows for valid inference using modern imputation methods.

Missing Not at Random (MNAR) occurs when missingness depends on unobserved values themselves. Consider patients dropping out of a depression study due to worsening symptoms — the unobserved depression scores directly influence the probability of missingness. MNAR presents the most challenging scenario and typically requires sensitivity analyses.

Ref:- https://medium.com/@saehwanpark/building-a-modern-mice-imputer-the-missing-data-challenge-in-research-3ec70fea7bdd

```r
qchat_missing <- is.na(asd_clean$qchat_10_score)

chisq.test(                                     # Checking whether missingness differs by ASD traits with Chi-square test
  table(qchat_missing, asd_clean$asd_traits)
)
```

```r
fisher.test(
  table(
    is.na(asd_clean$qchat_10_score), # Checking whether missingness differs by ASD traits with Fisher test
    asd_clean$asd_traits
  )
)
```

QChat score missingness is systematic by ASD traits. Could be MAR.

```r
srs_missing <- is.na(asd_clean$social_responsiveness_scale)

chisq.test(                              # Checking whether missingness differs by ASD traits with Chi-square test
  table(srs_missing, asd_clean$asd_traits)
)
```

```r
fisher.test(
  table(srs_missing, asd_clean$asd_traits) # Checking whether missingness differs by ASD traits with Fisher test
    asd_clean$asd_traits
)
```

SRS score missingness is not systematic by ASD traits. Could be MCAR or maybe the 9 values are not sufficient to draw conclusion as also evidenced by the Chi-square test warning.

```r
behaviour_missing <- is.na(asd_clean$social_behavioural_issues)

chisq.test(                                     # Checking whether missingness differs by ASD traits with Chi-square test
  table(behaviour_missing, asd_clean$asd_traits)
)
```

```r
fisher.test(
  table(
    is.na(asd_clean$social_behavioural_issues), # Checking whether missingness differs by ASD traits with Fisher test
    asd_clean$asd_traits
    asd_clean$asd_traits
  )
)
```

Social Behavioral Issues missingness is not systematic by ASD traits. Could be MCAR or maybe the 14 values are not sufficient to draw conclusion.
